# basic-world-map

Run from the **repository root** so `lib` imports work (`jupyter lab` with cwd at repo root).

```python
import importlib.util

from lib.paths import map_dir, repo_root

MAP_SLUG = "basic-world-map"
MAP_DIR = map_dir(MAP_SLUG)
OUTPUT_DIR = MAP_DIR / "output"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Repo:", repo_root())
print("Map:", MAP_DIR)

main_path = MAP_DIR / "main.py"
spec = importlib.util.spec_from_file_location("basic_world_map_main", main_path)
main_mod = importlib.util.module_from_spec(spec)
assert spec.loader is not None
spec.loader.exec_module(main_mod)

output_path = OUTPUT_DIR / "world_map.png"
main_mod.draw_world_map(output_path)
print(f"Wrote {output_path}")
```
