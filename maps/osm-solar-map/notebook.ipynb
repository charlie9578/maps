# osm-solar-map

Run from the **repository root** (`jupyter lab` with cwd at repo root).

**Kernel:** **Maps (.venv)**. Install map deps first if needed: `pip install -r maps/osm-solar-map/requirements.txt`

```python
import importlib.util

from lib.paths import map_dir, repo_root

MAP_SLUG = "osm-solar-map"
MAP_DIR = map_dir(MAP_SLUG)
OUTPUT_DIR = MAP_DIR / "output"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Repo:", repo_root())
print("Map:", MAP_DIR)

main_path = MAP_DIR / "main.py"
module_name = MAP_SLUG.replace("-", "_") + "_main"
spec = importlib.util.spec_from_file_location(module_name, main_path)
main_mod = importlib.util.module_from_spec(spec)
assert spec.loader is not None
spec.loader.exec_module(main_mod)

out = main_mod.write_map(main_mod.OUTPUT_HTML, refresh_data=False)
print(f"Wrote {out}")
frame = main_mod.fetch_overpass(refresh=False)
print(f"Features: {len(frame):,}")
```
