This map is a Dash app; the recommended way to run it is from a terminal:

```bash
python maps/gem-wind-map/main.py
```


````python
import importlib.util

from lib.paths import map_dir, repo_root

MAP_SLUG = "gem-wind-map"
MAP_DIR = map_dir(MAP_SLUG)

print("Repo:", repo_root())
print("Map:", MAP_DIR)

main_path = MAP_DIR / "main.py"
module_name = MAP_SLUG.replace("-", "_") + "_main"
spec = importlib.util.spec_from_file_location(module_name, main_path)
main_mod = importlib.util.module_from_spec(spec)
assert spec.loader is not None
spec.loader.exec_module(main_mod)

# Running Dash from inside a notebook is usually awkward (it blocks the kernel).
# Prefer starting it from a terminal as shown above.
# main_mod.main()
````
