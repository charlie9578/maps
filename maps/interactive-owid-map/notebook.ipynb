Run from the **repository root** so `lib` imports work (`jupyter lab` with cwd at repo root).

**Kernel:** choose **Maps (.venv)** (repo `.venv`). If it is missing, from the repo root:

```bash
.venv\Scripts\python.exe -m pip install -r maps/interactive-owid-map/requirements.txt
.venv\Scripts\python.exe -m ipykernel install --user --name maps --display-name "Maps (.venv)"
```

Then reload the notebook and pick **Maps (.venv)** from the kernel picker.

````python
import importlib.util

from lib.paths import map_dir, repo_root

MAP_SLUG = "interactive-owid-map"
MAP_DIR = map_dir(MAP_SLUG)
OUTPUT_DIR = MAP_DIR / "output"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Repo:", repo_root())
print("Map:", MAP_DIR)

main_path = MAP_DIR / "main.py"
spec = importlib.util.spec_from_file_location("interactive_owid_map_main", main_path)
main_mod = importlib.util.module_from_spec(spec)
assert spec.loader is not None
spec.loader.exec_module(main_mod)

output_path = OUTPUT_DIR / "co2_per_capita_map.html"
main_mod.write_map(output_path)
print(f"Wrote {output_path}")
output_path
````
