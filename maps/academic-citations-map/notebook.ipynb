````python
import sys

from lib.paths import map_dir, repo_root

MAP_SLUG = "academic-citations-map"
MAP_DIR = map_dir(MAP_SLUG)
if str(MAP_DIR) not in sys.path:
    sys.path.insert(0, str(MAP_DIR))

from data_processing import expand_citation_network, search_works
from viz import build_network_figure

print("Repo:", repo_root())
print("Map:", MAP_DIR)

from data_processing import pick_work

results = search_works("Penmanshiel wind farm data")
seed = pick_work(results, query="Penmanshiel wind farm data")
print(f"Seed: {seed.id} — {seed.display_name}")

network = expand_citation_network(seed.id, depth=1)
fig = build_network_figure(network)
out = MAP_DIR / "output" / "citation_network.html"
out.parent.mkdir(parents=True, exist_ok=True)
fig.write_html(out, include_plotlyjs="cdn")
print(f"Wrote {out} ({len(network.nodes)} nodes, {len(network.edges)} edges)")
````
